```python
#resizing the image 
import cv2
import matplotlib.pyplot as plt


img = cv2.imread(r"C:\Users\Anuradha\Downloads\flower.jpg")
resized = cv2.resize(img,(200,200))
plt.subplot(1,2,1)
plt.title("Resized Image")
plt.imshow(cv2.cvtColor(resized,cv2.COLOR_BGR2RGB)) #converting to rgb
plt.axis("off")

plt.subplot(1,2,2)
plt.title("Original Image")
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB)) #converting to rgb
plt.axis("off")
```
