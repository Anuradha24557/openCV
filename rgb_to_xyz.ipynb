```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\Anuradha\Downloads\flower.jpg")

rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
xyz = cv2.cvtColor(img,cv2.COLOR_BGR2XYZ)

plt.subplot(1,2,1)
plt.imshow(rgb)
plt.title("RGB IMAGE")
plt.axis("off")

plt.subplot(1,2,2)
plt.title("XYZ IMAGE")
plt.imshow(xyz)
plt.axis("off")
```
