```python
import cv2
import matplotlib.pyplot as plt

# Read images
img1 = cv2.imread(r"C:\Users\Anuradha\Downloads\catt.jpeg")
img2 = cv2.imread(r"C:\Users\Anuradha\Downloads\dogg.jpeg")

# Resizing image
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

# Blending images
alpha = 0.30
beta = 0.70

final_image = cv2.addWeighted(img1, alpha, img2, beta, 0)

# Converting BGR to RGB 
final_image_rgb = cv2.cvtColor(final_image, cv2.COLOR_BGR2RGB)


plt.imshow(final_image_rgb)
plt.title("Blended Image")
plt.axis("off")
plt.show()
```
