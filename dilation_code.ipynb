```python
#OpenCV CODE
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread(r"C:\Users\Anuradha\Downloads\3pp.jpg")

# Converting BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Kernel sizes
d1 = 0
d2 = 10
d3 = 20

# Kernel types
t1 = cv2.MORPH_RECT
t2 = cv2.MORPH_CROSS
t3 = cv2.MORPH_ELLIPSE

# Create kernels
kernel1 = cv2.getStructuringElement(t1, (2*d1+1, 2*d1+1))
kernel2 = cv2.getStructuringElement(t2, (2*d2+1, 2*d2+1))
kernel3 = cv2.getStructuringElement(t3, (2*d3+1, 2*d3+1))

# Apply dilation
dilate1 = cv2.dilate(image, kernel1)
dilate2 = cv2.dilate(image, kernel2)
dilate3 = cv2.dilate(image, kernel3)

# Converting to RGB
dilate1 = cv2.cvtColor(dilate1, cv2.COLOR_BGR2RGB)
dilate2 = cv2.cvtColor(dilate2, cv2.COLOR_BGR2RGB)
dilate3 = cv2.cvtColor(dilate3, cv2.COLOR_BGR2RGB)


plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(dilate1)
plt.title("Dilation - Rectangle")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(dilate2)
plt.title("Dilation - Cross")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(dilate3)
plt.title("Dilation - Ellipse")
plt.axis("off")


plt.show()
```
