```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\Anuradha\Downloads\flower.jpg")

yuv = cv2.cvtColor(img,cv2.COLOR_BGR2YUV)
rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

plt.subplot(1,2,1)
plt.imshow(yuv)
plt.title("YUV IMAGE")
plt.axis("off")

plt.subplot(1,2,2)
plt.title("RGB IMAGE")
plt.imshow(rgb)
plt.axis("off")
```
