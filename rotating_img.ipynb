```python
import cv2
from PIL import Image
from IPython.display import display, HTML

img = cv2.imread(r"C:\Users\Anuradha\Downloads\flower.jpg")

rows, cols = img.shape[:2]

matrix = cv2.getRotationMatrix2D((cols/2, rows/2), 45, 1)
rotated = cv2.warpAffine(img, matrix, (cols, rows))

rotated_img = Image.fromarray(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))


rotated_img = rotated_img.resize((300, 300))

display(HTML("<h3>Rotated Image</h3>"))
display(rotated_img)
```
