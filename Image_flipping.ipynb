```python
import cv2
import matplotlib.pyplot as plt 

img = cv2.imread(r"C:\Users\Anuradha\Downloads\flower.jpg")
flipped = cv2.flip(img,1)

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(flipped,cv2.COLOR_BGR2RGB))
plt.title("flipped")
plt.axis("off")

plt.show()
```
